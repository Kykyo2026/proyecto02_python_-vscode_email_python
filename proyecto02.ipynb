```python
import yfinance
```

```python
tiker= input("Escriba el código de la acción: ")
data= yfinance.download([tiker], period="6mo")
cierre= data.Close[tiker]
```

```python
cierre.plot()
```

## HACER LOS ANALISIS SOLICITADOS POR EL GESTOR

```python
cierre
```

```python
máxima= round(cierre.max(), 2)
mínimo= round(cierre.min(), 2)
valor_medio= round(cierre.mean(), 2)
```

```python
print(máxima)
print(mínimo)
print(valor_medio)
```
